<h3 style="text-align:left; color:#1E90FF;">
  3. Combining Well Log Data with Formation Tops
</h3>

Formation tops represent the depths at which different geological formations are encountered in the well. Integrating this with log data provides crucial context for petrophysical analysis.

```python
tops = pd.read_csv(r'C:\Users\hp\Coding World\PE-Petrophysical Analytics\Datasets\formation_tops.csv')

tops.head()
```

```python
merged_data = pd.merge(df, tops, left_on = 'DEPTH', right_on='TVD', how='left')

merged_data
```

```python
merged_data['FORMATION'] = merged_data['FORMATION'].fillna(method='ffill')

merged_data
```

<h3 style="text-align:left; color:#1E90FF;">
  4. Combining Core Data with Log Data
</h3>

Core data provides direct measurements of rock properties but is only available at specific depths. Combining it with continuous log data allows for calibration and improved interpretation.

```python
core_data = pd.read_csv(r'C:\Users\hp\Coding World\PE-Petrophysical Analytics\Notes\core_data.csv')

core_data.head()
```

```python
interpolated_logs = df.reindex(core_data['DEPTH']).interpolate()

interpolated_logs
```
